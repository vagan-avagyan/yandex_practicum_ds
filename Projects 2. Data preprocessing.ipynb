Для категориальных атрибутов () следующего рода проверку - оценим отношение кол-ва каждого из значений вышеперечисленных атрибутов к общему числу строк в рассматриваемой группе и в случае, если в обеих группах строк эти отношения будут приблизительно равны, то значит содержание рассматриваемых атрибутов никак не влияет на некорректное заполнение атрибута  `children`.

Так как подобные манипуляции понадобятся для проверки пропусков и в других атрибутах, то напишем функцию которая будет вызывается с заданными параметрами.

```python
value_counts_ratio(
    df=df,
    missing_filter_expr=(pl.col("children") == -1),
    columns_list=[
        "dob_years",
        "education",
        "family_status",
        "income_type"
    ]
)
```

```python
def value_counts_ration(missing_filter_expr, columns_list):
    """
    missing_filter_expr - выражение Polars для отбора строк с пропусками (например, pl.col('children') == -1)
    columns_list - список рассматриваемых атрибутов
    """
    curr_num = 0
    for curr_column in columns_list:
        curr_num += 1
        missing_group = df.filter(missing_filter_expr)
        filled_group = df.filter(~missing_filter_expr)
        missing_group_val_cnt = missing_group[curr_column].value_counts()
        filled_group_val_cnt = filled_group[curr_column].value_counts()
        missing_group_shape = missing_group.height
        filled_group_shape = filled_group.height
        print(f"{curr_num}. Атрибут '{curr_column}' - отношение кол-во каждого из его значений к общему числу строк, %:")
        print('')
        print(f"a. Группа строк с пропусками:\n{missing_group_val_cnt.with_columns((pl.col('count') / missing_group_shape * 100).alias('%'))}")
        print('')
        print(f"b. Группа строк заполненными значениями:\n{filled_group_val_cnt.with_columns((pl.col('count') / filled_group_shape * 100).alias('%'))}")
        print('='*100)
        print('')
```

Рассматриваемое отношение для обеих групп строк примерно одинаковые.

В группе строк с корректно заполненными значениями у некоторых атрибутов есть значения, которые отсутствуют в группе строк с пропусками, но так как их доля невелика, то они не влияют на появление пропусков - следовательно данные пропуски можно отнести к типу "Полностью случайные"


```python
value_counts_ration(pl.col('children') == -1, ['gender', 'income_type', 'education_id', 'family_status_id'])
```

Так как в данных приведена информация о клиентах банка, то давайте попробуем их объединить в подгруппы по некоторыми признакам. Для каждой подгруппы посчитаем медианное значения кол-ва детей (т.к. медина устойчива к выбросам) и используем эти значения для корректировки некорректно заполненных строк.

Для того, чтобы определить перечень атрибутов, по которым стоит произвести группировку, поступим следующим образом - возьмем каждый рассматриваемый атрибут в отдельности, произведем группировку и посчитаем медианное значения для каждой полученной подгруппы. Используем группу строк с корректно заполненными значениями в атрибуте  `children`.

Рассматривать будем все те же атрибуты:
- `dob_years`
- `education_id`
- `family_status_id`
- `gender`
- `income_type`

В этом случае также напишем функцию, для дальнейшего применения.

```python
def groupby_target_column_val(target_column, missing_filter_expr, columns_list):
    curr_num = 0
    filled_df = df.filter(~missing_filter_expr)
    for curr_column in columns_list:
        curr_num += 1
        med = filled_df.group_by(curr_column).agg(pl.col(target_column).median().round(2))
        print(f"{curr_num}. Атрибут {curr_column}\n{med}")
        print('='*100)
        print('')
```

Глядя на полученные результаты, можно сделать вывод, что группировку данных для заполнения пропущенных значения следует проводить только по атрибуту `dob_years`.


```python
groupby_target_column_val('children', pl.col('children') == -1, ['dob_years', 'education_id',
                                                                  'family_status_id', 'gender',
                                                                  'income_type'])
```

#### Атрибут `days_employed`

Перед тем как определить тип пропусков и способ его заполнения приведем значения в данном атрибуте в единый формат.

1. В атрибуте есть положительные и отрицательные числа
2. Название атрибута предполагает содержание кол-ва дней, но:
    - только отрицательные числа приведены в днях
    - положительные в часах

```python
# Отрицательные — в днях, положительные — переводим из часов в дни (делим на 24), затем abs и round
df = df.with_columns(
    pl.when(pl.col('days_employed') < 0)
    .then(pl.col('days_employed').abs())
    .otherwise(pl.col('days_employed') / 24)
    .abs()
    .round(2)
    .alias('days_employed')
)
df['days_employed']
```

Сильной вариативности медианного значения `dob_years` в атрибутах `family_status_id` и `education_id` - нет. Поэтому не будет учитывать их при группировке данных.

Чем больше атрибутов, по которым производится группировка, тем больше подгрупп данных с пустыми медийными значениями.

```python
# Сгруппируем данные по выбранным атрибутам и посчитаем медиану для каждой подгруппы
temp_group_df = (
    df.select(['gender', 'income_type', 'dob_years', 'days_employed'])
    .group_by(['gender', 'income_type', 'dob_years'])
    .agg(pl.col('days_employed').median().alias('days_employed_median'))
)
temp_group_df.filter(pl.col('days_employed_median').is_null()).shape
```

Так как атрибутов по которым проводится группировка много, то не будем писать функцию на подобии `dob_years_group_children_cnt`, а поступим иначе.

```python
# Соединяем df с temp_group_df по ключам (left join)
df = df.join(temp_group_df, on=['gender', 'income_type', 'dob_years'], how='left')
# Заполняем пропуски в days_employed медианой по группе
df = df.with_columns(
    pl.when(pl.col('days_employed').is_null())
    .then(pl.col('days_employed_median'))
    .otherwise(pl.col('days_employed'))
    .alias('days_employed')
)
```

### Категоризация данных

#### Атрибут `dob_years`

Первый атриубут по которому можно провести категоризацию данных - это атрибут `dob_years`. Для этого воспользуемся функцие из теоретической части тернажера.

```python
def age_group(age):
    """
    Возвращает возрастную группу по значению возраста age, используя правила:
    - 'дети', если age <= 18 лет;
    - 'взрослые', если age от 19 до 64;
    - 'пенсионеры' — от 65 и старше.
    """
    
    if age <= 18:
        return 'дети'
    if age <= 64:
        return 'взрослые'
    return 'пенсионеры'
```

```python
df = df.with_columns(
    pl.col('dob_years').map_elements(age_group, return_dtype=pl.Utf8).alias('dob_years_group')
)
df.head(10)
```

#### Атрибут `purpose`

Аналогичным образом можем поступить и с атрибутом `purpose`.

Для разнообразия проделаем действия аналогичные функции `age_group`, только при помощи `lambda`-функции 

```python
def purpose_to_group(x):
    if x is None:
        return None
    if 'коммерческая недвижимость' in x:
        return 'коммерческая недвижимость'
    if 'недвижимость' in x or 'жилье' in x:
        return 'недвижимость'
    if 'автомобиль' in x:
        return 'автомобиль'
    if 'образование' in x:
        return 'образование'
    if 'свадьба' in x:
        return 'свадьба'
    return None

df = df.with_columns(
    pl.col('purpose').map_elements(purpose_to_group, return_dtype=pl.Utf8).alias('purpose_group')
)
df.head(10)
```

#### Атрибут `total_income`

Категоризируем клиентов банка по уровню дохода.

Пороги выбраны субъективно, так как в зависимости от информации в интернете данные пороги разные.

При реальном случае решения подобной задачи необходимо узнать, есть ли у банка аналитика по данному вопросу и брать именно те пороги, которые определил сам банк в ходе анализа.

```python
# Категории по доходу: до 13890 (МРОТ), до 60000, до 150000, выше
df = df.with_columns(
    pl.when(pl.col('total_income') <= 13890).then(pl.lit('ниже прожиточного минимуму'))
    .when(pl.col('total_income') <= 60000).then(pl.lit('низкий уровень дохода'))
    .when(pl.col('total_income') <= 150000).then(pl.lit('средний уровень дохода'))
    .otherwise(pl.lit('высокий уровень дохода'))
    .alias('total_income_group')
)
df.head(10)
```
